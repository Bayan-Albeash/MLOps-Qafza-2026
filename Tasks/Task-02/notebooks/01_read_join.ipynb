```python
import pandas as pd
from sqlalchemy import create_engine

DB_URL = "postgresql+psycopg2://postgres:postgres@localhost:5432/olist"

engine = create_engine(DB_URL)

print("Connection object created successfully")
```

```python
from sqlalchemy import inspect

inspector = inspect(engine)

tables = inspector.get_table_names()

print("Tables in database:")
for table in tables:
    print("-", table)
```

```python
dataframes = {}

for table in tables:
    df = pd.read_sql(f"SELECT * FROM {table}", engine)
    dataframes[table] = df
    
    print(f"{table}: {len(df)} rows")
```

```python
for table, df in dataframes.items():
    print("=" * 60)
    print("TABLE:", table)
    print("Shape:", df.shape)
    print("Columns:")
    print(df.columns.tolist())
    print("\nFirst 2 rows:")
    display(df.head(2))
```

```python
customers = dataframes["customers"]

print("Total rows:", len(customers))
print("Unique customer_id:", customers["customer_id"].nunique())
print("Unique customer_unique_id:", customers["customer_unique_id"].nunique())

print("Duplicate customer_id:",
      customers["customer_id"].duplicated().sum())

print("Duplicate customer_unique_id:",
      customers["customer_unique_id"].duplicated().sum())
```

```python
summary = []

for table, df in dataframes.items():
    summary.append({
        "table": table,
        "rows": df.shape[0],
        "columns": df.shape[1],
        "duplicate_rows": df.duplicated().sum()
    })

summary_df = pd.DataFrame(summary)

display(summary_df)
```

```python
key_checks = {
    "customers": ["customer_id", "customer_unique_id"],
    "orders": ["order_id", "customer_id"],
    "products": ["product_id"],
    "sellers": ["seller_id"],
    "order_items": ["order_id", "product_id", "seller_id"],
    "order_payments": ["order_id"],
    "order_reviews": ["review_id", "order_id"],
    "geolocation": ["geolocation_zip_code_prefix"],
    "product_category_translation": ["product_category_name"]
}

for table, columns in key_checks.items():
    df = dataframes[table]

    print("=" * 60)
    print("TABLE:", table)

    for col in columns:
        print(
            col,
            "| unique:", df[col].nunique(),
            "| duplicates:", df[col].duplicated().sum()
        )
```

```python
for table in ["orders", "order_items", "order_payments"]:
    print("=" * 60)
    print("TABLE:", table)
    print(dataframes[table].columns.tolist())
    display(dataframes[table].head(3))
```

```python
display(dataframes["order_payments"].head(5))
```

```python
payments = dataframes["order_payments"]

repeated_order = payments[
    payments["order_id"].duplicated(keep=False)
].sort_values("order_id")

display(repeated_order.head(10))
```

```python
payments_agg = (
    payments
    .groupby("order_id")
    .agg(
        total_payment_value=("payment_value", "sum"),
        payment_records=("payment_sequential", "count"),
        max_installments=("payment_installments", "max"),
        payment_types=("payment_type", lambda x: ", ".join(sorted(set(x.dropna()))))
    )
    .reset_index()
)

display(payments_agg.head())
```

```python
print("Rows before aggregation:", len(payments))
print("Rows after aggregation:", len(payments_agg))
print("Unique order_id after aggregation:", payments_agg["order_id"].nunique())
print("Duplicate order_id after aggregation:", payments_agg["order_id"].duplicated().sum())
```

```python
items = dataframes["order_items"]

print(items.columns.tolist())
display(items.head(5))
```

```python
items_agg = (
    items
    .groupby("order_id")
    .agg(
        item_count=("order_item_id", "count"),
        total_item_price=("price", "sum"),
        total_freight_value=("freight_value", "sum"),
        unique_products=("product_id", "nunique"),
        unique_sellers=("seller_id", "nunique")
    )
    .reset_index()
)

display(items_agg.head())
```

```python
orders = dataframes["orders"]

print(orders.columns.tolist())
display(orders.head())
```

```python
customers = dataframes["customers"]

ml_table = orders.merge(
    customers,
    on="customer_id",
    how="left"
)

print("Rows after customers join:", len(ml_table))
print("Unique order_id:", ml_table["order_id"].nunique())
print("Duplicate order_id:", ml_table["order_id"].duplicated().sum())

display(ml_table.head())
```

```python
ml_table = ml_table.merge(
    items_agg,
    on="order_id",
    how="left"
)

print("Rows after items join:", len(ml_table))
print("Unique order_id:", ml_table["order_id"].nunique())
print("Duplicate order_id:", ml_table["order_id"].duplicated().sum())
```

```python
ml_table = ml_table.merge(
    payments_agg,
    on="order_id",
    how="left"
)

print("Rows after payments join:", len(ml_table))
print("Unique order_id:", ml_table["order_id"].nunique())
print("Duplicate order_id:", ml_table["order_id"].duplicated().sum())
```

```python
print("Missing item_count:", ml_table["item_count"].isna().sum())
print("Missing total_item_price:", ml_table["total_item_price"].isna().sum())
print("Missing total_payment_value:", ml_table["total_payment_value"].isna().sum())
print("Missing payment_records:", ml_table["payment_records"].isna().sum())
```

```python
missing_items = ml_table[ml_table["item_count"].isna()]

print(missing_items["order_status"].value_counts())
```

```python
missing_payments = ml_table[ml_table["total_payment_value"].isna()]

print(missing_payments["order_status"].value_counts())
```

```python
print("Final ML table shape:", ml_table.shape)
print("Unique order_id:", ml_table["order_id"].nunique())
print("Duplicate order_id:", ml_table["order_id"].duplicated().sum())

display(ml_table.head())
```
