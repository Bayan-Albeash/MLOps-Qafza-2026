```python

import numpy as np
import pandas as pd
from pathlib import Path
import joblib

from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

ARTIFACTS_DIR = Path("../artifacts")
```

```python

X_train = np.load(ARTIFACTS_DIR / "X_train_processed.npy")
X_val = np.load(ARTIFACTS_DIR / "X_val_processed.npy")
X_test = np.load(ARTIFACTS_DIR / "X_test_processed.npy")

y_train = pd.read_csv(ARTIFACTS_DIR / "y_train.csv").squeeze("columns")
y_val = pd.read_csv(ARTIFACTS_DIR / "y_val.csv").squeeze("columns")
y_test = pd.read_csv(ARTIFACTS_DIR / "y_test.csv").squeeze("columns")

print("X_train:", X_train.shape)
print("X_val:", X_val.shape)
print("X_test:", X_test.shape)

print("y_train:", y_train.shape)
print("y_val:", y_val.shape)
print("y_test:", y_test.shape)
```

```python
# Late = 1, On-time = 0

label_mapping = {
    "On-time": 0,
    "Late": 1
}

y_train_bin = y_train.map(label_mapping)
y_val_bin = y_val.map(label_mapping)
y_test_bin = y_test.map(label_mapping)

print(y_train_bin.value_counts())
```

```python

def evaluate_model(model, X, y_true, dataset_name):
    y_pred = model.predict(X)

    if hasattr(model, "predict_proba"):
        y_prob = model.predict_proba(X)[:, 1]
    else:
        y_prob = None

    accuracy = accuracy_score(y_true, y_pred)
    precision = precision_score(y_true, y_pred, zero_division=0)
    recall = recall_score(y_true, y_pred, zero_division=0)
    f1 = f1_score(y_true, y_pred, zero_division=0)

    print(f"\n===== {dataset_name} =====")
    print("Accuracy:", round(accuracy, 4))
    print("Precision:", round(precision, 4))
    print("Recall:", round(recall, 4))
    print("F1-score:", round(f1, 4))

    if y_prob is not None:
        roc_auc = roc_auc_score(y_true, y_prob)
        print("ROC-AUC:", round(roc_auc, 4))
    else:
        roc_auc = None

    print("\nConfusion Matrix:")
    print(confusion_matrix(y_true, y_pred))

    print("\nClassification Report:")
    print(classification_report(y_true, y_pred, zero_division=0))

    return {
        "dataset": dataset_name,
        "accuracy": accuracy,
        "precision": precision,
        "recall": recall,
        "f1": f1,
        "roc_auc": roc_auc
    }
```

```python

baseline_model = DummyClassifier(
    strategy="most_frequent"
)

baseline_model.fit(X_train, y_train_bin)

baseline_val_results = evaluate_model(
    baseline_model,
    X_val,
    y_val_bin,
    "Baseline - Validation"
)
```

```python

logreg_model = LogisticRegression(
    max_iter=1000,
    class_weight="balanced",
    random_state=42
)

logreg_model.fit(X_train, y_train_bin)

logreg_val_results = evaluate_model(
    logreg_model,
    X_val,
    y_val_bin,
    "Logistic Regression - Validation"
)
```

```python

C_values = [0.01, 0.1, 1, 10]

tuning_results = []

for C in C_values:
    model = LogisticRegression(
        C=C,
        max_iter=1000,
        class_weight="balanced",
        random_state=42
    )

    model.fit(X_train, y_train_bin)

    y_val_pred = model.predict(X_val)
    y_val_prob = model.predict_proba(X_val)[:, 1]

    result = {
        "C": C,
        "precision": precision_score(y_val_bin, y_val_pred, zero_division=0),
        "recall": recall_score(y_val_bin, y_val_pred, zero_division=0),
        "f1": f1_score(y_val_bin, y_val_pred, zero_division=0),
        "roc_auc": roc_auc_score(y_val_bin, y_val_prob)
    }

    tuning_results.append(result)

tuning_df = pd.DataFrame(tuning_results)

display(tuning_df.sort_values("f1", ascending=False))
```

```python

best_row = tuning_df.sort_values(
    "f1",
    ascending=False
).iloc[0]

best_C = best_row["C"]

print("Best C:", best_C)
print("Best Validation F1:", best_row["f1"])
print("Best Validation ROC-AUC:", best_row["roc_auc"])
```

```python

final_model = LogisticRegression(
    C=best_C,
    max_iter=1000,
    class_weight="balanced",
    random_state=42
)

final_model.fit(X_train, y_train_bin)
```

```python
# Use test set once at the end

final_test_results = evaluate_model(
    final_model,
    X_test,
    y_test_bin,
    "Final Model - Test"
)
```

```python

results_summary = pd.DataFrame([
    baseline_val_results,
    logreg_val_results,
    final_test_results
])

display(results_summary)
```

```python

model_path = ARTIFACTS_DIR / "logistic_regression_model.joblib"

joblib.dump(final_model, model_path)

print("Saved model to:", model_path)
```

```python

results_path = ARTIFACTS_DIR / "model_results_summary.csv"

results_summary.to_csv(
    results_path,
    index=False
)

print("Saved results summary to:", results_path)
```

## Training and Evaluation Summary

- A majority-class DummyClassifier was used as the baseline.
- Logistic Regression was selected as the initial machine learning model.
- Class imbalance was handled using `class_weight="balanced"`.
- Hyperparameter tuning was performed only on the validation set.
- F1-score was used as the main tuning metric because the Late class is the minority class.
- Precision, Recall, F1-score, ROC-AUC, Accuracy, and the Confusion Matrix were evaluated.
- The test set was used only once after model selection.
- The final trained model and results summary were saved as artifacts.

### Why accuracy is not enough

The dataset is imbalanced, with the Late class representing only a small portion of all orders. A model could achieve high accuracy by predicting most orders as On-time while failing to detect Late orders. Therefore, Recall, Precision, F1-score, and ROC-AUC are more informative metrics for this problem.
